## Data Preparation: From 60 Compressed Files to One Working CSV

The course distributes ExtraSensory as **60 separate `.features_labels.csv.gz` files**, one per anonymized participant. Each compressed file contains the same sensor and label columns but does not contain a user-ID column; the ID is stored in its filename. Before the analysis, the files must therefore be read, assigned user identifiers, reduced to the columns relevant to this project, and concatenated row-wise.

The code below shows the complete reproducible process. To avoid re-reading hundreds of megabytes every time the notebook runs, it creates `project_df.csv` only when that file is absent. If the combined CSV already exists, the notebook loads it directly. This cached file is an intermediate working dataset derived entirely from the 60 course files.

```python
# Locate the folder of 60 compressed participant files and the cached combined CSV.
raw_folder_candidates = [Path('ExtraSensory'), Path('finalproject_codex/ExtraSensory')]
raw_folder = next((folder for folder in raw_folder_candidates if folder.exists()), None)
if raw_folder is None:
    raise FileNotFoundError('The ExtraSensory folder was not found.')

raw_files = sorted(raw_folder.glob('*.features_labels.csv.gz'))
assert len(raw_files) == 60, f'Expected 60 compressed user files, found {len(raw_files)}.'

important_columns = [
    'timestamp',
    'raw_acc:magnitude_stats:mean',
    'raw_acc:magnitude_stats:std',
    'raw_acc:magnitude_stats:percentile50',
    'raw_acc:magnitude_stats:time_entropy',
    'raw_acc:magnitude_spectrum:spectral_entropy',
    'raw_acc:magnitude_autocorrelation:normalized_ac',
    'proc_gyro:magnitude_stats:std',
    'location:num_valid_updates',
    'location:diameter',
    'location:max_speed',
    'audio_naive:mfcc0:mean',
    'audio_properties:max_abs_value',
    'discrete:app_state:is_active',
    'lf_measurements:battery_level',
    'lf_measurements:screen_brightness',
    'discrete:time_of_day:between0and6',
    'discrete:time_of_day:between21and3',
    'label:SLEEPING']

combined_csv_candidates = [Path('project_df.csv'), Path('finalproject_codex/project_df.csv')]
combined_csv = next((path for path in combined_csv_candidates if path.exists()), combined_csv_candidates[0])

if combined_csv.exists():
    # Reuse the previously generated working file for faster repeated execution.
    raw_df = pd.read_csv(combined_csv)
    preparation_message = f'Loaded existing {combined_csv}.'
else:
    # Read each gzip file, recover its user ID from the filename, and concatenate all users.
    user_frames = []
    for user_number, file_path in enumerate(raw_files, start=1):
        user_df = pd.read_csv(
            file_path, compression='gzip', usecols=important_columns
        )
        user_df.insert(0, 'uuid', f'No{user_number}')
        user_frames.append(user_df)

    raw_df = pd.concat(user_frames, ignore_index=True)
    local_datetime = (
        pd.to_datetime(raw_df['timestamp'], unit='s', utc=True)
        .dt.tz_convert('America/Los_Angeles')
    )
    raw_df.insert(2, 'datetime', local_datetime.astype(str))
    raw_df.insert(3, 'hour', local_datetime.dt.hour)
    raw_df.to_csv(combined_csv, index=False)
    preparation_message = f'Created {combined_csv} from 60 compressed user files.'

print(preparation_message)
print(f'Compressed files found: {len(raw_files)}')
print(f'Combined working-data shape: {raw_df.shape}')
print(f'Anonymized users: {raw_df["uuid"].nunique()}')
raw_df.head()
```

## Step 1: Introduction

The ExtraSensory dataset contains smartphone and smartwatch measurements collected from 60 people during everyday life. Approximately 20 seconds of sensor readings were summarized in each roughly one-minute observation. The measurements include motion, location, audio, phone state, battery level, screen brightness, and context labels reported by participants.

This project asks: **How does passive smartphone activity differ when a user is sleeping versus not sleeping, and can ordinary phone measurements predict whether the user is sleeping?** Sleep recognition may reduce the burden of manual self-reporting and support wellness applications, but missing labels and differences between users must be considered.

After concatenating the 60 participant files, the dataset contains **377,346 rows**. Each row represents one sensor-summary window for one anonymized user. The columns relevant to this investigation are:

- `uuid`: anonymized user identifier (nominal).
- `timestamp`: Unix time for the observation.
- `label:SLEEPING`: 1 when sleeping, 0 when not sleeping, and missing when the state was not reported.
- `raw_acc:magnitude_stats:std`: standard deviation of phone acceleration magnitude.
- `discrete:app_state:is_active`: whether an app was active.
- `lf_measurements:battery_level`: phone battery level.
- `lf_measurements:screen_brightness`: screen brightness.
- `hour` and `is_night`: local-time features created during cleaning.

## Step 3: Assessment of Missingness

### MNAR Analysis

`label:SLEEPING` is plausibly **MNAR** because its missingness may depend on the unobserved true value: a participant who is asleep cannot report that state at the time, and tired participants may ignore a prompt. This conclusion comes from reasoning about the data-generating process, not merely from patterns in the DataFrame. Additional information about prompt times, response delays, charging state, and whether the screen was on could help explain the missingness, making an MAR explanation more plausible.

### Missingness Dependency

The selected column is `label:SLEEPING`. For each observed binary feature, the test statistic is the absolute difference in the proportion equal to 1 between rows where the sleep label is missing and rows where it is observed. For a binary distribution, this statistic equals TVD.

For both tests, the null hypothesis says that sleep-label missingness does not depend on the observed feature; the alternative says that it does depend on the feature. We use 2,000 permutations and pre-specify **α = 0.001** because two related time variables are tested on a very large dataset.

```python
# Define a permutation test for dependence between response missingness and a binary feature.
def missingness_permutation_test(data, feature, repetitions=2000, seed=42):
    test_df = data[['label:SLEEPING', feature]].dropna(subset=[feature])
    values = test_df[feature].to_numpy()
    missing = test_df['label:SLEEPING'].isna().to_numpy()

    def statistic(group):
        return abs(values[group].mean() - values[~group].mean())

    observed = statistic(missing)
    rng = np.random.default_rng(seed)
    simulated = np.empty(repetitions)
    for repetition in range(repetitions):
        simulated[repetition] = statistic(rng.permutation(missing))

    p_value = (np.count_nonzero(simulated >= observed) + 1) / (repetitions + 1)
    return observed, p_value, simulated
```

```python
# Test a broader night indicator and the narrower 9 PM–3 AM indicator.
night_test = missingness_permutation_test(project_df, 'is_night')
evening_col = 'discrete:time_of_day:between21and3'
evening_test = missingness_permutation_test(project_df, evening_col)

missingness_results = pd.DataFrame({
    'Feature': ['is_night', evening_col],
    'Observed TVD': [night_test[0], evening_test[0]],
    'p-value': [night_test[1], evening_test[1]]
})
display(missingness_results)

fig_missing = px.histogram(
    x=night_test[2], nbins=35,
    title='Null Distribution: Sleep-Label Missingness and Night Status',
    labels={'x': 'TVD Between Night-Status Distributions', 'count': 'Permutations'},
    color_discrete_sequence=['#72B7B2']
)
fig_missing.add_vline(
    x=night_test[0], line_color='#E45756', line_width=3,
    annotation_text='Observed TVD', annotation_position='top right'
)
fig_missing.show()
```

At α = 0.001, we reject the null for `is_night`, so the missingness of `label:SLEEPING` depends on the broader night period. We fail to reject the null for `between21and3`, so this test does not provide sufficient evidence that missingness depends on the narrower 9 PM–3 AM indicator. Failing to reject does not prove independence or MCAR.

## Step 4: Hypothesis Testing

The project question suggests testing whether the phone moves less during sleep.

- **Null hypothesis:** In the population represented by these observations, mean phone acceleration standard deviation is the same while users are sleeping and not sleeping. Any observed difference is due to random chance under the null.
- **Alternative hypothesis:** Mean phone acceleration standard deviation is lower while users are sleeping.
- **Test statistic:** mean acceleration standard deviation for not-sleeping observations minus the mean for sleeping observations. Large positive values favor the alternative.
- **Method and significance level:** a one-sided permutation test with 1,000 repetitions and α = 0.01. A difference in means directly measures the directional claim about average movement.

```python
# Shuffle sleep labels to generate the null distribution of the difference in means.
acc_values = sleep_acc[acc_col].to_numpy()
sleep_labels = sleep_acc['label:SLEEPING'].astype(int).to_numpy()

def acceleration_mean_difference(labels):
    return acc_values[labels == 0].mean() - acc_values[labels == 1].mean()

observed_difference = acceleration_mean_difference(sleep_labels)
rng = np.random.default_rng(RANDOM_SEED)
repetitions = 1000
null_differences = np.empty(repetitions)
for repetition in range(repetitions):
    null_differences[repetition] = acceleration_mean_difference(
        rng.permutation(sleep_labels)
    )

hypothesis_p_value = (
    np.count_nonzero(null_differences >= observed_difference) + 1
) / (repetitions + 1)

print(f'Observed mean difference: {observed_difference:.6f}')
print(f'Permutation p-value: {hypothesis_p_value:.6f}')
print('Decision:', 'Reject H0' if hypothesis_p_value < 0.01 else 'Fail to reject H0')

fig_hypothesis = px.histogram(
    x=null_differences, nbins=40,
    title='Null Distribution of the Acceleration Mean Difference',
    labels={'x': 'Mean(Not Sleeping) − Mean(Sleeping)', 'count': 'Permutations'},
    color_discrete_sequence=['#4C78A8']
)
fig_hypothesis.add_vline(
    x=observed_difference, line_color='#E45756', line_width=3,
    annotation_text='Observed statistic', annotation_position='top right'
)
fig_hypothesis.show()
```

The p-value is below 0.01, so we reject the null hypothesis. The observed data provide strong evidence that mean phone acceleration variability is lower during reported sleep. Because this is observational data rather than a randomized experiment, the result identifies an association and does not prove that sleep causes the difference.

## Step 5: Framing a Prediction Problem

The prediction task is to predict `label:SLEEPING` from passive measurements available in the same one-minute window. This is **binary classification**, with sleeping as the positive class. Rows with a missing response are excluded because their correct class is unknown.

The primary evaluation metric is the **F1-score for sleeping**. Sleeping is the minority class, so accuracy can look high even when many sleeping observations are missed. F1-score balances precision and recall. Accuracy and the classification report are secondary diagnostics.

At the time of prediction, local time, acceleration variability, and current app state are already available. Other self-reported labels are excluded because they may not be available and could leak the response. Because many rows come from the same person, users—not individual rows—are split between training and test sets. The test set therefore measures generalization to unseen users.

## Step 6: Baseline Model

Before evaluating the required baseline model, we define a simple **time-only benchmark**: predict sleeping from 9 PM through 5:59 AM and predict not sleeping at all other hours. This fixed rule represents how well a reasonable clock-based guess performs without sensor measurements. It is only a comparison benchmark, not the official baseline.

The official baseline uses three features: local `hour` and `raw_acc:magnitude_stats:std` are **two quantitative features**, while `discrete:app_state:is_active` is **one binary nominal feature**. The latter is already represented as 0/1, so it needs no further encoding. Importantly, acceleration and app state are two columns selected from the original data; hour is derived directly from the original timestamp. Missing feature values are replaced with the training-set median. A depth-3, class-balanced decision-tree classifier is intentionally simple and interpretable. Both the imputer and classifier are contained in one sklearn Pipeline, so the imputer is fitted only on training data and leakage is avoided.

```python
# Split by uuid so no participant appears in both the training and test sets.
baseline_features = [
    'hour', 'raw_acc:magnitude_stats:std', 'discrete:app_state:is_active'
]
model_df = project_df.dropna(subset=['label:SLEEPING']).copy()
X = model_df[baseline_features]
y = model_df['label:SLEEPING'].astype(int)
groups = model_df['uuid']

splitter = GroupShuffleSplit(n_splits=1, test_size=0.20, random_state=RANDOM_SEED)
train_positions, test_positions = next(splitter.split(X, y, groups=groups))
X_train, X_test = X.iloc[train_positions], X.iloc[test_positions]
y_train, y_test = y.iloc[train_positions], y.iloc[test_positions]
train_users = groups.iloc[train_positions]
test_users = groups.iloc[test_positions]

assert set(train_users).isdisjoint(set(test_users))
print(f'Training rows/users: {len(X_train):,} / {train_users.nunique()}')
print(f'Test rows/users: {len(X_test):,} / {test_users.nunique()}')
print(f'Training sleep rate: {y_train.mean():.3f}')
print(f'Test sleep rate: {y_test.mean():.3f}')
```

```python
# Fit all baseline transformations and the classifier in one Pipeline.
baseline_model = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('classifier', DecisionTreeClassifier(
        max_depth=3, class_weight='balanced', random_state=RANDOM_SEED
    ))
])
baseline_model.fit(X_train, y_train)

# Apply the fixed time-only rule to the same unseen test users.
time_only_predictions = (
    (X_test['hour'] >= 21) | (X_test['hour'] < 6)
).astype(int)

train_predictions = baseline_model.predict(X_train)
test_predictions = baseline_model.predict(X_test)
baseline_metrics = pd.DataFrame({
    'Split': ['Train', 'Test (unseen users)'],
    'F1 for Sleeping': [
        f1_score(y_train, train_predictions),
        f1_score(y_test, test_predictions)
    ],
    'Accuracy': [
        accuracy_score(y_train, train_predictions),
        accuracy_score(y_test, test_predictions)
    ]
})
display(baseline_metrics)

# Compare both approaches on exactly the same unseen observations.
model_comparison = pd.DataFrame({
    'Model': ['Time-only benchmark', 'Three-feature baseline'],
    'Accuracy': [
        accuracy_score(y_test, time_only_predictions),
        accuracy_score(y_test, test_predictions)
    ],
    'Precision for Sleeping': [
        precision_score(y_test, time_only_predictions),
        precision_score(y_test, test_predictions)
    ],
    'Recall for Sleeping': [
        recall_score(y_test, time_only_predictions),
        recall_score(y_test, test_predictions)
    ],
    'F1 for Sleeping': [
        f1_score(y_test, time_only_predictions),
        f1_score(y_test, test_predictions)
    ]
})
display(model_comparison)
print(classification_report(
    y_test, test_predictions, target_names=['Not Sleeping', 'Sleeping']
))
```

On the same unseen users, the time-only benchmark achieves an F1-score of about **0.62** and accuracy of about **0.77**. The three-feature baseline improves these values to an F1-score of about **0.76** and accuracy of about **0.86**. It has sleeping-class precision of about **0.69** and recall of about **0.83**. This is a useful baseline because it substantially beats the clock rule and finds most sleeping records, although roughly 31% of its positive predictions are false positives. Its lower test F1 than training F1 also shows some loss in generalization across people. There remains room to improve the representation of cyclic time and the phone's motion state.

## Step 7: Final Model

The final model keeps exactly the same training and test users as the baseline. Before tuning, we choose to search random-forest `max_depth` and `min_samples_leaf`. Maximum depth controls how complicated each tree can become, while minimum leaf size regularizes the model by preventing rules based on very small groups of records. A random forest can capture nonlinear interactions between time, movement, and phone state while averaging many decision trees to make predictions more stable.

The model engineers four new features for reasons tied to the data-generating process:

1. `hour_sin` and `hour_cos` represent time cyclically, so 11 PM and midnight are close to one another.
2. `log_acc_std` reduces the extreme right skew in acceleration variability and preserves differences among low-motion observations, which are especially relevant to sleep.
3. `log_gyro_std` provides a second measure of phone motion and is log-transformed for the same skewness reason.

App-active state, battery level, and screen brightness are retained as low-sensitivity indicators of current phone use. All engineered features, median imputation, and classification are contained in one Pipeline. We use three-fold `GroupKFold` cross-validation on training users only, select the parameter combination with the highest mean sleeping-class F1-score, and evaluate the selected model once on the untouched test users.

```python
# Engineer cyclic-time and log-motion features inside the model Pipeline.
class SleepFeatureEngineer(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None):
        return self

    def transform(self, X):
        engineered = pd.DataFrame(index=X.index)
        engineered['hour_sin'] = np.sin(2 * np.pi * X['hour'] / 24)
        engineered['hour_cos'] = np.cos(2 * np.pi * X['hour'] / 24)
        engineered['log_acc_std'] = np.log1p(
            X['raw_acc:magnitude_stats:std'].clip(lower=0)
        )
        engineered['log_gyro_std'] = np.log1p(
            X['proc_gyro:magnitude_stats:std'].clip(lower=0)
        )
        engineered['app_active'] = X['discrete:app_state:is_active']
        engineered['battery_level'] = X['lf_measurements:battery_level']
        engineered['screen_brightness'] = X['lf_measurements:screen_brightness']
        return engineered


final_features = [
    'hour', 'raw_acc:magnitude_stats:std',
    'proc_gyro:magnitude_stats:std', 'discrete:app_state:is_active',
    'lf_measurements:battery_level', 'lf_measurements:screen_brightness'
]
X_final = model_df[final_features]
X_final_train = X_final.iloc[train_positions]
X_final_test = X_final.iloc[test_positions]

final_pipeline = Pipeline([
    ('features', SleepFeatureEngineer()),
    ('imputer', SimpleImputer(strategy='median')),
    ('classifier', RandomForestClassifier(
        n_estimators=120, class_weight='balanced',
        random_state=RANDOM_SEED, n_jobs=-1
    ))
])
```

```python
# Tune only on training users with group-aware folds, using sleeping-class F1.
parameter_grid = {
    'classifier__max_depth': [8, 12],
    'classifier__min_samples_leaf': [10, 50]
}
group_cv = GroupKFold(n_splits=3)
grid_search = GridSearchCV(
    final_pipeline,
    parameter_grid,
    scoring='f1',
    cv=group_cv,
    n_jobs=-1,
    return_train_score=True
)
grid_search.fit(X_final_train, y_train, groups=train_users)

cv_results = pd.DataFrame(grid_search.cv_results_)[[
    'param_classifier__max_depth',
    'param_classifier__min_samples_leaf',
    'mean_test_score', 'std_test_score', 'mean_train_score'
]].rename(columns={
    'mean_test_score': 'Mean Validation F1',
    'std_test_score': 'Validation F1 SD',
    'mean_train_score': 'Mean Training F1'
})
display(cv_results.sort_values('Mean Validation F1', ascending=False))
print('Best parameters:', grid_search.best_params_)
print(f'Best mean validation F1: {grid_search.best_score_:.3f}')
```

```python
# Evaluate the selected, refitted model once on the unchanged test users.
final_model = grid_search.best_estimator_
final_predictions = final_model.predict(X_final_test)

final_comparison = pd.DataFrame({
    'Model': ['Baseline decision tree', 'Final random forest'],
    'Accuracy': [
        accuracy_score(y_test, test_predictions),
        accuracy_score(y_test, final_predictions)
    ],
    'Precision for Sleeping': [
        precision_score(y_test, test_predictions),
        precision_score(y_test, final_predictions)
    ],
    'Recall for Sleeping': [
        recall_score(y_test, test_predictions),
        recall_score(y_test, final_predictions)
    ],
    'F1 for Sleeping': [
        f1_score(y_test, test_predictions),
        f1_score(y_test, final_predictions)
    ]
})
display(final_comparison)

confusion = pd.crosstab(
    pd.Series(y_test.to_numpy(), name='Actual').map({0: 'Not Sleeping', 1: 'Sleeping'}),
    pd.Series(final_predictions, name='Predicted').map({0: 'Not Sleeping', 1: 'Sleeping'})
)
fig_confusion = px.imshow(
    confusion,
    text_auto=True,
    color_continuous_scale='Blues',
    title='Final Model Confusion Matrix on Unseen Users',
    labels={'x': 'Predicted Status', 'y': 'Actual Status', 'color': 'Records'}
)
fig_confusion.show()
```

The best cross-validation parameters are `max_depth=8` and `min_samples_leaf=50`, with a mean validation F1-score of about **0.816**. After refitting on all training users, the final model reaches a sleeping-class F1-score of about **0.771** and accuracy of about **0.874** on unseen test users. This improves on the baseline's F1-score of **0.755** and accuracy of **0.864** using the identical split. Precision rises from about **0.695** to **0.715**, while recall also rises slightly from about **0.827** to **0.836**. The improvement is modest but consistent across both the primary metric and accuracy.

## Step 8: Fairness Analysis

We ask whether the final model performs worse at recognizing actual sleep during the **day** than during the **night** on the unchanged test set. Night is defined as 9 PM through 5:59 AM. This comparison matters because a sleep detector that mainly learns conventional nighttime schedules may miss naps or sleep among users with nontraditional schedules.

The evaluation metric is **recall for sleeping**, the proportion of actual sleeping records identified by the model. Recall is appropriate because the relevant harm is failing to recognize sleep.

- **Null hypothesis:** The final model's sleeping-class recall is the same during night and day; the observed difference is due to random assignment of time-period labels.
- **Alternative hypothesis:** Sleeping-class recall is lower during the day than at night.
- **Test statistic:** night recall minus day recall. Large positive values favor the alternative.
- **Method:** a one-sided permutation test with 5,000 repetitions at α = 0.05. The fitted model, actual outcomes, and predictions remain fixed; only day/night labels are shuffled.

```python
# Compare recall by time period and shuffle group labels under the fairness null.
fairness_df = pd.DataFrame({
    'actual': y_test.to_numpy(),
    'predicted': final_predictions,
    'is_night': (
        (X_final_test['hour'] >= 21) | (X_final_test['hour'] < 6)
    ).to_numpy()
})

sleeping_records = fairness_df[fairness_df['actual'] == 1]
night_recall = sleeping_records.loc[
    sleeping_records['is_night'], 'predicted'
].mean()
day_recall = sleeping_records.loc[
    ~sleeping_records['is_night'], 'predicted'
].mean()
observed_recall_gap = night_recall - day_recall

rng = np.random.default_rng(RANDOM_SEED)
repetitions = 5000
simulated_recall_gaps = np.empty(repetitions)
period_labels = fairness_df['is_night'].to_numpy()
actual_sleep = fairness_df['actual'].to_numpy() == 1
sleep_predictions = fairness_df.loc[actual_sleep, 'predicted'].to_numpy()

for repetition in range(repetitions):
    shuffled_period = rng.permutation(period_labels)[actual_sleep]
    simulated_recall_gaps[repetition] = (
        sleep_predictions[shuffled_period].mean()
        - sleep_predictions[~shuffled_period].mean()
    )

fairness_p_value = (
    np.count_nonzero(simulated_recall_gaps >= observed_recall_gap) + 1
) / (repetitions + 1)

display(pd.DataFrame({
    'Period': ['Night', 'Day'],
    'Sleeping Recall': [night_recall, day_recall],
    'Actual Sleeping Records': [
        sleeping_records['is_night'].sum(),
        (~sleeping_records['is_night']).sum()
    ]
}))
print(f'Observed recall difference (night - day): {observed_recall_gap:.3f}')
print(f'One-sided permutation p-value: {fairness_p_value:.6f}')

fig_fairness = px.histogram(
    x=simulated_recall_gaps,
    nbins=40,
    title='Fairness Test: Null Distribution of the Recall Difference',
    labels={'x': 'Night Recall − Day Recall', 'count': 'Permutations'},
    color_discrete_sequence=['#72B7B2']
)
fig_fairness.add_vline(
    x=observed_recall_gap,
    line_color='#E45756',
    line_width=3,
    annotation_text='Observed difference',
    annotation_position='top right'
)
fig_fairness.show()
```

The final model's sleeping-class recall is about **0.907 at night** and **0.632 during the day**, an observed difference of about **0.276**. The one-sided permutation p-value is about **0.0002**, below α = 0.05, so we reject the null hypothesis. The test provides strong evidence that the model performs worse at recognizing actual daytime sleep than nighttime sleep. This does not mean the model is unfair in every possible sense; it identifies a specific recall disparity for these time-period groups and suggests that additional examples or features representing daytime sleep may be needed.
